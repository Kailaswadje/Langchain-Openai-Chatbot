```python
# create a function and pass the prompt given by the user and returns the AI response  
# clear prompt given to BOT as to how it shld behave and respond for unwanted questions
def chat_bot(prompt):
    response = client.chat.completions.create(model = "gpt-4o-mini",
                                              messages=[
                                                  {
                                                      "role":"system",
                                                      "content":"""You are known to be polite and helpful AI BOT.
                                                                  Act as a teaching assistant for a Data Science course 
                                                                  working with an edtech. Your job here is to help students
                                                                  resolve their doubts regarding specific data science topics.
                                                                  If the doubt is not relevant to data science, you can politely 
                                                                  ask user another doubt and don't give answer for that particular question."""
                                                  },
                                                  {"role":"user", 
                                                   "content":prompt}])
    return response.choices[0].message.content
```

# Adding Memory

```python
# history = []  - This creates a list to store conversation memory ,It will hold: user messages / assistant replies
# So the bot can “remember” past chat.
#history.append({"role":"user", "content": prompt}) -Every time user types something: it is saved into history 
# + history in messages line add  prev messages ie the past conversations 
# saving assitant reply  - stores AI RESPONSE also 

history = []
def chat_bot(prompt):
    history.append({"role":"user", "content":prompt})
    response = client.chat.completions.create(model = "gpt-4o-mini",
                                              messages=[
                                                  {
                                                      "role":"system",
                                                      "content":"""You are known to be polite and helpful AI BOT.
                                                                  Act as a teaching assistant for a Data Science course 
                                                                  working with an edtech. Your job here is to help students
                                                                  resolve their doubts regarding specific data science topics.
                                                                  If the doubt is not relevant to data science, you can politely 
                                                                  ask user another doubt and don't give answer for that particular question."""
                                                  }] + history )
    history.append({"role":"assistant", "content": response.choices[0].message.content})
    return response.choices[0].message.content
```

# FLIPPED INTERACTION PATTERN

```python
#Flipped interaction = when AI stops only answering and instead drives the learning by 
# asking, guiding, or reversing roles.
#This code is building a chatbot that:
#uses a prompt template
#connects to an LLM (chatgpt)
#stores conversation history automatically
#manages memory per session

#ChatPromptTemplate → builds structured prompts
#ChatMessageHistory → stores chat memory
#RunnableWithMessageHistory → connects memory + model + session system

# These are LangChain’s “chat + memory framework tools”
```

```python
from langchain_core.prompts import ChatPromptTemplate
from langchain_community.chat_message_histories import ChatMessageHistory
from langchain_core.runnables.history import RunnableWithMessageHistory
# This defines the format of what the model sees. Structure:"placeholder", "{chat_history}"/ injects past conversation automatically
#"human", "{input}"  current user question

prompt = ChatPromptTemplate.from_messages(
    [
        ("placeholder", "{chat_history}"),
        ("human", "{input}"),
    ]
)

#What this means: Pipe operator | = “connect steps”  So: prompt formats input/  output goes to chatgpt /chatgpt generates response

llm_chain = (prompt | chatgpt)

# history is a memory object which store user messages and AI responses
history = ChatMessageHistory()

### here prompt + llm chain + session id ( returns same session storage)
#every session uses this history so chat remembers previous messages

conv_chain = RunnableWithMessageHistory(
    llm_chain,
    lambda session_id : history,
    input_messages_key = "input",
    hisotry_messages_key = "chat_history",
)
```

```python
prompt_input = """I would like you to ask me questions to build a detailed project plan
            for a data science project.
            Ask me questions until you have enough information to build this project plan.
            You can ask questions about the problem, resources, data, and any other relevant questions.
            Ask me one question at a time. Let’s start with the first question.
         """
response = conv_chain.invoke({"input": prompt_input},
                             {"configurable": {"session_id": "unused"}})
display(Markdown(response.content))
```
